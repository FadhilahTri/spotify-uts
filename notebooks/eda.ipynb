```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load dataset
df = pd.read_csv('spotify_data clean.csv')

# Cek bentuk data
print("Jumlah baris dan kolom:", df.shape)
print("\n5 data pertama:")
df.head()
```

```python
# Informasi detail dataset
print("=== INFO DATASET ===")
print(df.info())

print("\n=== STATISTIK DESKRIPTIF ===")
df.describe()
```

```python
# Cek missing values
print("=== MISSING VALUES ===")
print(df.isnull().sum())

# Handle missing values - isi genre yang kosong dengan 'unknown'
df['artist_genres'] = df['artist_genres'].fillna('unknown')

# Konversi kolom explicit dari True/False ke 1/0
df['explicit'] = df['explicit'].astype(int)

# Konversi album_release_date ke tahun saja
df['release_year'] = pd.to_datetime(df['album_release_date'], errors='coerce').dt.year
df['release_year'] = df['release_year'].fillna(df['release_year'].median()).astype(int)

# Encoding album_type
df['album_type_encoded'] = df['album_type'].map({
    'album': 0, 'single': 1, 'compilation': 2
}).fillna(0).astype(int)

print("\n=== DATA SETELAH PREPROCESSING ===")
print("Missing values tersisa:")
print(df.isnull().sum())
print("\nShape:", df.shape)
```

```python
# Visualisasi EDA
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('EDA - Spotify Track Popularity', fontsize=16)

# 1. Distribusi track_popularity
axes[0,0].hist(df['track_popularity'], bins=30, color='#1DB954', edgecolor='white')
axes[0,0].set_title('Distribusi Track Popularity')
axes[0,0].set_xlabel('Popularity Score')
axes[0,0].set_ylabel('Jumlah Lagu')

# 2. Popularity vs Artist Popularity
axes[0,1].scatter(df['artist_popularity'], df['track_popularity'], 
                  alpha=0.3, color='#1DB954', s=10)
axes[0,1].set_title('Track vs Artist Popularity')
axes[0,1].set_xlabel('Artist Popularity')
axes[0,1].set_ylabel('Track Popularity')

# 3. Popularity per tahun
yearly = df.groupby('release_year')['track_popularity'].mean()
axes[1,0].plot(yearly.index, yearly.values, color='#1DB954', linewidth=2)
axes[1,0].set_title('Rata-rata Popularity per Tahun')
axes[1,0].set_xlabel('Tahun')
axes[1,0].set_ylabel('Avg Popularity')

# 4. Korelasi heatmap
numeric_cols = ['track_popularity', 'artist_popularity', 
                'artist_followers', 'track_duration_min',
                'album_total_tracks', 'release_year']
corr = df[numeric_cols].corr()
sns.heatmap(corr, annot=True, fmt='.2f', cmap='Greens', ax=axes[1,1])
axes[1,1].set_title('Korelasi Antar Fitur')

plt.tight_layout()
plt.savefig('eda_visualisasi.png', dpi=150, bbox_inches='tight')
plt.show()
print("Visualisasi berhasil disimpan!")
```

```python
# Simpan dataset yang sudah bersih
df.to_csv('spotify_preprocessed.csv', index=False)
print("✅ Dataset berhasil disimpan sebagai 'spotify_preprocessed.csv'")
print(f"Total data: {df.shape[0]} baris, {df.shape[1]} kolom")
print("\nKolom siap untuk modeling:")
feature_cols = ['artist_popularity', 'artist_followers', 'track_duration_min',
                'track_number', 'album_total_tracks', 'explicit', 
                'release_year', 'album_type_encoded']
print(feature_cols)
print("\nTarget variabel: track_popularity")
```
